```python
#FACTORIAL
n=int(input("selecciona un numero entero para calcular su factorial:"))
k=1
for i in range (2,n+1):
    k*=i
print("el factorial de",n,"es:
",k)
```

```python
#CREAR UNA LISTA DE ALUMNOS
lista=[]
def agregar():
    nombre=str(input("Nombre completo del alumno:"))
    lista.append(nombre)
    return (0)

def eliminar(lista):
    for i in range (len(lista)):
        print("[",i,"]\t",lista[i])
    selecc=int(input("Seleccione registro a eliminar:"))
    lista.pop(selecc)
    return lista
    
print("Sistema de registro de alumnos.")
accion=1
while accion!=0:
    lista.sort()
    if lista==[]:
        print("\nLa lista está vacía!")
    else: 
        print("\nLa lista actual es:")
        for i in lista:
            print(i)
    accion=int(input("Seleccione una opcion:\n1) Agregar nuevo alumno\n2) Eliminar registro\n0) Salir\n"))
    if accion==1:
        agregar()
    elif accion==2:
        lista=eliminar(lista)
    else:
        print("opcion no valida")


        
```
